```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

# Add root to path
sys.path.insert(0, str(Path.cwd().parent))

from src.data.load import load_raw_data

# Load 50k sample
df = load_raw_data(mode="test")

# 1. Check Severity Distribution
print("Severity Counts:\n", df['Severity'].value_counts(normalize=True) * 100)

# 2. Plot it
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Severity', palette='viridis')
plt.title('Distribution of Accident Severity (50k Sample)')
plt.ylabel('Count')
plt.show()
```
